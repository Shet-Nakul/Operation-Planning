```python
import json
import yaml
import websockets
import json

with open('multi.json') as f:
    multi = json.load(f)

with open("optimizer_config.yml") as f:
    config = yaml.safe_load(f)

with open('single.json') as f:
    single = json.load(f)
```

```python
async def call_planning_ws():
    uri = "ws://localhost:8000/planning"
    async with websockets.connect(uri) as websocket:
        payload = {
            "data": multi,
            "config": config
        }

        await websocket.send(json.dumps(payload))

        try:
            while True:
                response = await websocket.recv()
                print("Server:", response)
        except websockets.exceptions.ConnectionClosed:
            print("Connection closed by server")
    return response

result = await call_planning_ws()
json.loads(json.loads(result)["result"])
```
