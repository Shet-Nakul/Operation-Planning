```python
import websockets
import asyncio
import json
import numpy as np

# Make sure the server is running first:
# uvicorn services.app:app --host 127.0.0.1 --port 8000

class NumpyEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, np.integer):
            return int(obj)
        if isinstance(obj, np.floating):
            return float(obj)
        if isinstance(obj, np.ndarray):
            return obj.tolist()
        return super().default(obj)

async def test_websocket():
    uri = "ws://127.0.0.1:8000/scheduling/"
    async with websockets.connect(uri) as ws:
        payload = json.dumps({"data": data, "config": config}, cls=NumpyEncoder)
        await ws.send(payload)

        while True:
            msg = json.loads(await ws.recv())
            if msg["status"] == "progress":
                print(f"Progress: {msg['data']}")
            elif msg["status"] == "completed":
                print(f"Completed")
                break
            elif msg["status"] == "error":
                print(f"Error: {msg['message']}")
                break
    return msg
msg = await test_websocket()
```
